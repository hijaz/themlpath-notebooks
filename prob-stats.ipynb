## Probability

**Probability:**
Probability is a measure of the likelihood of an event occurring. It's a number between 0 and 1, where 0 indicates an impossible event and 1 indicates a certain event. Probability is often denoted by the symbol P.

Example: When rolling a fair six-sided die, the probability of rolling a 3 is 1/6 because there is one favorable outcome (rolling a 3) out of six possible outcomes.

**Complement of Probability:**
The complement of the probability of an event A (denoted as A') is the probability of event A not occurring. It's the probability of everything else happening except event A.

Example: If the probability of rolling a 3 on a die is 1/6, then the probability of not rolling a 3 (complement) is 1 - 1/6 = 5/6.

3. Joint and Disjoint (Mutually Exclusive) Events:

Joint Events: Joint events are events that occur together. The probability of joint events happening is denoted as P(A and B), where A and B are two different events.
Example: When flipping a coin and rolling a die, the joint event of getting heads on the coin and rolling a 4 on the die has a probability of 1/2 * 1/6 = 1/12.

Disjoint (Mutually Exclusive) Events: Disjoint events are events that cannot happen simultaneously. If one event occurs, the other cannot occur at the same time.
Example: When drawing a card from a standard deck, the events "drawing a heart" and "drawing a spade" are disjoint because they cannot happen together in a single draw.

4. Sum of Probabilities for Joint and Disjoint Events:

Joint Events: The sum of probabilities of joint events is calculated by adding the probabilities of each event occurring together.
Example:
Suppose we have a fair six-sided die, and we want to find the probability of rolling a 3 on the first roll and a 5 on the second roll.

Probability of rolling a 3 on the first roll: P(3) = 1/6
Probability of rolling a 5 on the second roll: P(5) = 1/6
The probability of rolling a 3 on the first roll and a 5 on the second roll (joint event) is:

P(3 and 5) = P(3) * P(5) = (1/6) * (1/6) = 1/36

Disjoint (Mutually Exclusive) Events: The sum of probabilities of disjoint events is calculated by adding the probabilities of each individual event.
Example: Drawing a card from a standard deck, the probability of drawing a heart or a spade (disjoint events) is P(heart or spade) = P(heart) + P(spade) = 1/4 + 1/4 = 1/2.
